```python
import numpy as np
from scipy import stats
```

```python
field_A = [1.2, 0.8, 1.5, 1.9, 1.1, 0.7, 1.4, 2.2, 1.3, 1.0, 1.6, 0.9, 1.1, 1.8, 0.6, 1.5, 1.3, 1.7, 1.2, 2.0, 0.9, 1.4];
field_B = [1.5, 1.7, 1.3, 1.9, 1.6, 2.0, 1.4, 1.8, 1.5, 1.2, 1.7, 2.1, 1.6, 1.4, 1.9, 1.8, 1.5, 1.3, 1.7, 1.6, 1.2, 1.8, 1.5, 1.4];
```

```python
t_stats, p_value = stats.ttest_ind(field_A, field_B, equal_var = False);
```

```python
t_stats
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print("I will reject the null hypothesis")
else:
    print("I will accept the null hypothesis")
```
