```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from scipy.stats import norm
```

```python
sample = [87, 89, 91, 93, 95, 96, 98, 99, 101, 102,103, 104, 105, 106, 106, 107, 107, 108, 109, 109,110, 
          110, 111, 111, 112, 112, 113, 114, 115, 116,117, 118, 119, 120, 122]
population_mean = 100
population_std = 15
sample_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - population_mean)/ (population_std / np.sqrt(n))
z_score
```

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("I will reject the null hypothesis")
else:
    print("I will accept the null hypothesis")
```
