```python
df.columns = df.columns.str.upper()
```

```python
df["SOLD PRICE($)"] = pd.to_numeric(df["SOLD PRICE($)"], errors="coerce")
df["Expensive"] = df["SOLD PRICE($)"].apply(
    lambda x: "YES" if  pd.notna(x) and x >= 15 else "NO"
)
```

```python
df
```

```python
df["Expensive"].value_counts()
```

```python
df["SOLD PRICE(CR.)"] = pd.to_numeric(df["SOLD PRICE(CR.)"], errors="coerce")
df["Expe. IN $"] = df["SOLD PRICE(CR.)"].apply(
    lambda y: "YES" if pd.notna(y) and y >= 4 else "NO"
)
```
