```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0,10,15)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("y=x^2")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x**2,y)
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(y*2, x)
```

**Object Oriented way**

```python
fig = plt.figure(figsize = (20,12),dpi=100)
axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2, 0.6, 0.3, 0.4])
axis2.plot(y,x)
```

```python
Cricketers = ["Virat", "Abd", "Rohit", "Sooryavanshi", "Smith"]

Match1 = [80, 45, 100, 65, 95]
Match2 = [60, 75, 50, 85, 70]
Match3 = [55, 90, 70, 60, 88]
Match4 = [100, 56, 41, 25, 10]
Match5 = [54, 152, 73, 56, 52]

bar_width = 0.15

x = np.arange(len(Cricketers))

plt.figure(figsize = (10,6))

plt.bar(x, Match1, width=bar_width, label="Match 1")
plt.bar(x + bar_width, Match2, width = bar_width, label="Match 2")
plt.bar(x + 2*bar_width, Match3, width = bar_width, label="Match 3")
plt.bar(x + 3*bar_width, Match4, width = bar_width, label="Match 4")
plt.bar(x + 4*bar_width, Match5, width = bar_width, label="Match 5")

plt.xticks(x + 2*bar_width, Cricketers)

plt.title("Runs Scored by Cricketers in Five Matches")
plt.xlabel("Players")
plt.ylabel("Runs")

plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

**Pie Chart**

```python
values = [20, 50, 64, 52, 99]
plt.pie(values)
plt.show()
```

```python
label = ["Apple", "Banana", "Orange", "Mango", "Grapes"]
colors = ["Red", "Yellow", "Orange", "purple", "Green"]
explode = [0,1,0,0,0]
plt.pie(values, labels=label, autopct="%1.1f%%", colors=colors, explode=explode, shadow=True, startangle=90)
plt.show()
```

```python
years = [2019, 2020, 2021, 2022, 2023]
sales = [120, 150, 180, 210, 250]

plt.figure(figsize=(10,6), dpi=120)

plt.plot(
    years, 
    sales,

    color="royalblue",

    linestyle = '--',

    linewidth = 3,
    marker = "o",
    markersize = 10,
    markerfacecolor="yellow",
    markeredgecolor="black",
    label="Sales"
)

plt.title(
    "Comapany Sales Analysis",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Years",
    fontsize=14,
    fontweight="bold"
)

plt.ylabel(
    "Sales (in Lakhs)",
    fontsize=14,
    fontweight="bold"
)

plt.grid(
    axis="both",
    linestyle=":",
    linewidth=1,
    alpha=0.7
)

plt.xlim(2018, 2024)
plt.ylim(100,270)

plt.xticks(rotation=0, fontsize=12)
plt.yticks(fontsize=12)

plt.legend(
    loc="upper left",
    fontsize = 12
)

plt.show()
```

```python
CGPA = [7.1, 7.4, 7.8, 8.0, 8.3, 8.5]
Semester = [1, 2, 3, 4, 5, 6]

plt.figure(figsize=(10,6), dpi=150)

plt.plot(
    Semester,
    CGPA,    

    color = "black",

    linestyle = "--",
    linewidth = 4,
    marker = "*",
    markersize=18,

    markerfacecolor = "green",
    markeredgecolor = "White",

    label="Semester"
)

plt.title(
    "Student CGPS Across Semesters",
    fontsize=15,
    fontweight="bold"
)

plt.ylabel(
    "Student CGPA",
    fontsize=12,
    fontweight="bold"
)

plt.xlabel(
    "Semester",
    fontsize=12,
    fontweight="bold"
)

plt.grid(
    axis="both",
    linestyle=":",
    linewidth=1.5,
    alpha=0.8
)

plt.xlim(0.5, 6.5)
plt.ylim(7,9)

plt.xticks(rotation=0, fontsize=12)
plt.yticks(fontsize=12)

plt.legend(
    loc="upper right",
    fontsize = 14
)

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug"]
units = [320, 290, 350, 410, 470, 520, 560, 530]

plt.figure(figsize=(10,6), dpi=150)

plt.plot(
    months,
    units,

    color = "orange",

    linestyle="-.",
    linewidth=3,

    marker = "D",
    markersize = 10,

    markerfacecolor = "cyan",
    markeredgecolor = "black",

    label="Electricity Units"
)

plt.title(
    "Monthly Electricity Consumption",
    fontsize = 16,
    fontweight = "bold"
)

plt.xlabel(
    "Month",
    fontsize = 12,
    fontweight = "bold"
)

plt.ylabel(
    "Units Consumed",
    fontsize = 12,
    fontweight = "bold"
)

plt.grid(
    axis = "both",
    linestyle = "--",
    linewidth = 1,
    alpha = 0.6
)

plt.ylim(250,600)

plt.xticks(rotation=45, fontsize=11)
plt.yticks(fontsize=11)

plt.legend(
    loc= "upper left",
    fontsize = 12
)

plt.show()
```

```python
import matplotlib.pyplot as plt

years = [2019,2020,2021,2022,2023]
sales = [120,150,180,210,250]

plt.figure(figsize=(8,5))

plt.plot(
    years,
    sales,
    marker="o",
    color="blue"
)

plt.annotate(

    "Highest Sales",

    xy=(2023,250),

    xytext=(2021.5,270),

    arrowprops=dict(
        arrowstyle="->"
    )
)

plt.title("Company Sales")

plt.xlabel("Year")

plt.ylabel("Sales")

plt.grid()

plt.show()
```
