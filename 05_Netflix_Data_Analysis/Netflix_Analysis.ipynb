In this section, we will derive new columns from the Netflix dataset, such as date-based features, duration information, genre information, and content indicators.

# EDA 1: Movies vs TV Shows

### Objective
Analyze the distribution of Movies and TV Shows available on Netflix.

```python
df_clone
```

```python
# Top 10 Countries with the most Netflix content
top_country = df_clone['country'].value_counts().head(10)
top_country
```

```python
# Find the most common genres available on Netflix.
top_genres = df_clone['primary_genre'].value_counts().head(15)
top_genres
```

```python
## Netflix Content Growth by Year
yearly = df_clone['year_added'].value_counts().sort_index()
yearly
```

```python
# Monthly Content Additions
month_order = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
monthly = df_clone['month_added'].value_counts().reindex(month_order)
monthly
```

```python
# Content Added by Weekday
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
weekday = df_clone['weekday_added'].value_counts().reindex(weekday_order)
weekday
```

```python
# Movie Duration Distribution
movies = df_clone[df_clone["type"] == "Movie"]
movies['duration_value'].describe()
```

```python
# Top Director
top_director = df_clone['director'].value_counts().head(15)
top_director
```

```python
# Top Actors
actors = (
    df_clone['cast'].str.split(',').explode().str.strip()
)

top_actor = actors.value_counts().head(15)
top_actor
```
