```python
import zipfile
import os

zip_path = "house-prices-advanced-regression-techniques.zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("data")

print("ZIP file extracted successfully!")
print(os.listdir("data"))
```

```python
train_path = "data/train.csv"
test_path = "data/test.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Datasets load successfully!")
print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)
```

```python
missing_values = train_df.isnull().sum()

missing_values = missing_values[missing_values > 0]

print("Columns with missing values:", len(missing_values))
print(missing_values.sort_values(ascending=False))
```

```python
missing_values = train_df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing_Count": missing_values,
    "Missing_Percentage": (missing_values / len(train_df)) * 100
})

missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_summary.round(2)
```

```python
with open("data/data_description.txt", "r", encoding="utf-8") as file:
    description = file.read()

columns_to_check = [
    "PoolQC",
    "MiscFeature",
    "Alley",
    "Fence",
    "FireplaceQu",
    "GarageType",
    "BsmtQual",
    "MasVnrType",
    "LotFrontage",
    "Electrical"
]

for column in columns_to_check:
    position = description.find(column)

    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)

    if position != -1:
        print(description[position:position + 600])
    else:
        print("Description not found.")
```
